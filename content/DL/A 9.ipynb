```python
import pandas as pd
import numpy as np 

from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense,Dropout
from tensorflow.keras.utils import to_categorical
```

```python
x_train=pd.read_csv("C:\\Users\\admin\\Downloads\\dataset\\X_train.txt", sep=r"\s+",header=None)
x_test=pd.read_csv("C:\\Users\\admin\\Downloads\\dataset\\X_test.txt", sep=r"\s+",header=None)
y_train=pd.read_csv("C:\\Users\\admin\\Downloads\\dataset\\y_train.txt", sep=r"\s+",header=None)
y_test=pd.read_csv("C:\\Users\\admin\\Downloads\\dataset\\y_test.txt", sep=r"\s+",header=None)
```

```python
print("X_train shape:", x_train.shape)
print("X_test shape:", x_test.shape)
print("y_train shape:", y_train.shape)
print("y_test shape:", y_test.shape)
```

```python
x_train.head(5)
```

```python
y_train.head(5)
```

```python
y_train = to_categorical(
    y_train.values.ravel() - 1,
    num_classes=6
)

y_test = to_categorical(
    y_test.values.ravel() - 1,
    num_classes=6
)
```

```python
# Build Neural Network
model=Sequential([
    Dense(64, activation='relu', input_shape=(x_train.shape[1],)),
    Dropout(0.3),
    Dense(32,activation='relu'),
    Dropout(0.3),
    Dense(6,activation='softmax')
])
```

```python
model.summary()
```

```python
# Compile Model 

model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy'],
)
    
print("Model compiled successfully!")
```

```python
history=model.fit(
    x_train,
    y_train,
    epochs=10,
    batch_size=32,
    validation_split=0.2
)
```

```python
# Evaluate the model on test data

loss, accuracy = model.evaluate(x_test, y_test)

print("Test Loss:", loss)
print("Test Accuracy:", accuracy)
```

```python
# Make Predictions
y_pred = model.predict(x_test)
y_pred_classes = np.argmax(y_pred, axis=1)
```

```python
# Activity Names
activity_names = [
    "Walking",
    "Walking Upstairs",
    "Walking Downstairs",
    "Sitting",
    "Standing",
    "Laying"
]
```

```python
# Convert actual labels back to class numbers

y_true_classes = np.argmax(y_test, axis=1)

print("Actual labels:", y_true_classes[:10])
print("Predicted labels:", y_pred_classes[:10])
```

```python
# Classification Report

print("\nClassification Report:\n")

print(classification_report(
    y_true_classes,
    y_pred_classes,
    target_names=activity_names
))
```

```python
# Confusion Matrix

print("\nConfusion Matrix:\n")

print(
    confusion_matrix(
        y_true_classes,
        y_pred_classes
    )
)
```

```python
# Predict Sample Activity

sample = x_test.iloc[0].values.reshape(1, -1)

prediction = model.predict(sample)

predicted_class = np.argmax(prediction)
actual_class = y_true_classes[0]

print("\nActual Activity:", activity_names[actual_class])
print("Predicted Activity:", activity_names[predicted_class])
```
